```python
import math

def human_readable_size(size_bytes):
    if size_bytes == 0:
        return "0 B"
    units = ['B', 'KB', 'MB', 'GB', 'TB']
    unit_index = int(math.floor(math.log(size_bytes, 1024)))
    return f"{size_bytes / (1024 ** unit_index):.2f} {units[unit_index]}"

# Usage:
# human_readable_size(file_size)  → e.g., "15.30 MB"
```

```python
import os
from pathlib import Path
from itertools import islice

library_path = Path(os.getenv('LIBRARY_PATH'))

test =  os.walk(library_path,topdown=True)
file_list = []
file_data = {}

for root, dirs, files in islice(os.walk(library_path), 0, None):
    if files:
        file_list.extend([os.path.join(root, f) for f in files])

for file_path in file_list:
    relative_path = os.path.relpath(file_path, library_path)
    file_name = os.path.basename(file_path)
    file_type = os.path.splitext(file_name)[1].lower()  # Get file extension and convert to lowercase
    file_size = os.path.getsize(file_path) # Get file size in bytes
    file_modified_time = os.path.getmtime(file_path)
    file_data[relative_path] = {
        'name': file_name,
        'type': file_type,
        'size': file_size,
        'modified_time': file_modified_time
    }

```

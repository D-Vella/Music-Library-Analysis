```python
from dotenv import load_dotenv
import os

# Load environment variables from .env file
load_dotenv()

# Read the library path
library_path = os.getenv('LIBRARY_PATH', 'NOT_SET')
print(f'Library path: {library_path}')
```

```python
import os
from pathlib import Path

library_path = Path(os.getenv('LIBRARY_PATH')) / 'Alestorm' / "Captain Morgan's Revenge"
print(f'Library path: {library_path}')

if library_path.exists():
    files = os.listdir(library_path)
    print(f'Files in library path: {files}')
else:
    print(f'Path does not exist: {library_path}')
```

```python
audio_files = [f for f in files if f.endswith(('.mp3', '.flac', '.wav'))]
import mutagen, sys

audio = mutagen.File(library_path / audio_files[0])  # returns None if mutagen doesn't recognise the file
if audio is None:
    sys.exit("Not a recognised audio file")

print(f"Container : {type(audio).__name__}")
print(f"Duration  : {audio.info.length:.1f}s")
print(f"Bitrate   : {getattr(audio.info, 'bitrate', 'n/a')}")
print(f"Tag format: {type(audio.tags).__name__ if audio.tags else 'NO TAGS'}")
print("-" * 60)
for key, value in (audio.tags or {}).items():
    print(f"{key:<35} {str(value)[:80]}")  # truncate: cover art is binary

# FLAC keeps artwork outside the tags
if hasattr(audio, "pictures"):
    print(f"\nEmbedded pictures: {len(audio.pictures)}")
```
